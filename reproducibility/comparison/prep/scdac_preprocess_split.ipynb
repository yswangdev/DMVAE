```python
import pandas as pd
project_root = "/scratch/g/chlin/Yushu/scDAC/scDAC"   # contains data/sim1, data/sim2, ...
out_data_root = os.path.join(project_root, "data")

for i in range(1, 21):
    meta_path = os.path.join(out_data_root, f"simmeta_{i}.txt")
    labels = np.loadtxt(meta_path)  # 1 column
    labels = labels.astype(int)

    out_dir = os.path.join(out_data_root, f"4_clus/sim{i}")
    os.makedirs(out_dir, exist_ok=True)

    out_csv = os.path.join(out_dir, "label.csv")
    df = pd.DataFrame({"cell": np.arange(len(labels)), "label": labels})
    df.to_csv(out_csv, index=False)

    print(f"Wrote {out_csv} (n={len(labels)})")
```
